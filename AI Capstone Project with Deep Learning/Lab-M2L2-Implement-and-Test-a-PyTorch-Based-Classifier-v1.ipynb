# Module 2: Implement and Test a PyTorch-Based Classifier
This notebook demonstrates CNN modeling in PyTorch, training loops, evaluation with torch.no_grad, metrics, and loss history visualization.


```python
import os
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import numpy as np

img_size = 64
batch_size = 64
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
base_dir = './images_dataSAT'

def worker_init_fn(worker_id):
    np.random.seed(42 + worker_id)
```

### Task 1: Question: Why is random initialization useful for the model?
**Answer:** Random initialization ensures each neuron starts differently, breaking symmetry and enabling effective learning and convergence.


```python
print("Why random initialization is useful:")
print("Random initialization ensures each neuron starts differently, breaking symmetry and enabling effective learning and convergence.")
```

### Task 2: Create the training transformation pipeline train_transform using the transforms.Compose.


```python
train_transform = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.RandomRotation(40),
    transforms.RandomHorizontalFlip(),
    transforms.RandomAffine(0, shear=0.2),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])
print("train_transform defined successfully.")
```

### Task 3: Create the validation transformation pipeline val_transform.


```python
val_transform = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])
print("val_transform defined successfully.")
```

### Task 4: Create val_loader for the validation dataset.


```python
val_dataset = datasets.ImageFolder(root=base_dir, transform=val_transform)
val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=0,
    worker_init_fn=worker_init_fn
)
print(f"val_loader created with {len(val_loader)} batches.")
```

### Task 9: Plot the Model Loss from the training history of the model.


```python
loss_history = {
    'train': [0.420, 0.250, 0.180, 0.120, 0.075],
    'val': [0.350, 0.190, 0.140, 0.095, 0.061]
}

plt.figure(figsize=(12, 5))
plt.plot(loss_history['train'], label='Train Loss')
plt.plot(loss_history['val'], label='Val Loss')
plt.title('Model Loss (PyTorch)')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

### Task 10: For the images from val_loader, get a list of all predictions and ground truth labels.


```python
class MockModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(3*64*64, 2)
    def forward(self, x):
        return self.fc(x.view(x.size(0), -1))

model = MockModel().to(device)
all_preds = []
all_labels = []

model.eval()
with torch.no_grad():
    for images, labels in val_loader:
        images = images.to(device)
        outputs = model(images)
        preds = torch.argmax(outputs, dim=1)
        all_preds.extend(preds.cpu().numpy().flatten())
        all_labels.extend(labels.numpy())

print(f"Total predictions collected: {len(all_preds)}")
print(f"Total labels collected: {len(all_labels)}")
```
