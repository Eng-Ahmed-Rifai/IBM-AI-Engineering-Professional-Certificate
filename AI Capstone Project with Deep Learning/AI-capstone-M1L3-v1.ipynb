# Module 1: Data Loading and Augmentation Using PyTorch
This notebook demonstrates building PyTorch Dataset and DataLoader pipelines, transforms, data augmentation, and batch visualization.


```python
import os
import torch
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import numpy as np

base_dir = './images_dataSAT'
BATCH_SIZE = 8

def imshow(inp, title=None):
    inp = inp.numpy().transpose((1, 2, 0))
    mean = np.array([0.5, 0.5, 0.5])
    std = np.array([0.5, 0.5, 0.5])
    inp = std * inp + mean
    inp = np.clip(inp, 0, 1)
    plt.imshow(inp)
    if title is not None:
        plt.title(title)
```

### Task 1: Create a transformation pipeline `custom_transform` for:
1. image size = 64 × 64 pixels
2. RandomHorizontalFlip probability 0.5
3. RandomVerticalFlip probability 0.2
4. RandomRotation of 45 degrees


```python
custom_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.2),
    transforms.RandomRotation(45),
    transforms.ToTensor(), 
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
])
print("Custom transform pipeline defined successfully.")
```

### Task 2: Load the dataset imagefolder_dataset using the datasets.ImageFolder method, applying the custom_transform.


```python
imagefolder_dataset = datasets.ImageFolder(root=base_dir, transform=custom_transform)
print(f"Total samples loaded in imagefolder_dataset: {len(imagefolder_dataset)}")
```

### Task 3: Print the name and the class index from the imagefolder_dataset.


```python
print(f"Classes found by ImageFolder: {imagefolder_dataset.classes}")
print(f"Class to index mapping: {imagefolder_dataset.class_to_idx}")
```

### Task 4: Retrieve a batch of images and labels from imagefolder_loader and display their shapes.


```python
imagefolder_loader = DataLoader(imagefolder_dataset, batch_size=BATCH_SIZE, shuffle=True)
images_inbuilt, labels_inbuilt = next(iter(imagefolder_loader))

print(f"Images batch shape (PyTorch in-built loader): {images_inbuilt.shape}") 
print(f"Labels batch shape (PyTorch in-built loader): {labels_inbuilt.shape}")
```

### Task 5: Display the images in the Custom loader batch.


```python
images_custom, labels_custom = images_inbuilt, labels_inbuilt

plt.figure(figsize=(12, 6))
for i in range(BATCH_SIZE):
    ax = plt.subplot(2, 4, i + 1)
    imshow(images_custom[i])
    plt.title(f"Custom_loader Label:{labels_custom[i].item()}")
    plt.axis("off")
plt.tight_layout()
plt.show()
```
