# Module 1: Comparison of Memory-Based and Generator-Based Data Loading
This notebook covers data loading, directory traversal, memory vs generator loading, and geospatial image dataset inspection.


```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

extract_dir = '.'
base_dir = os.path.join(extract_dir, 'images_dataSAT')
dir_non_agri = os.path.join(base_dir, 'class_0_non_agri')
dir_agri = os.path.join(base_dir, 'class_1_agri')

non_agri_images = sorted([f for f in os.listdir(dir_non_agri) if f.endswith(('.jpg', '.jpeg', '.png'))])
agri_images = sorted([f for f in os.listdir(dir_agri) if f.endswith(('.jpg', '.jpeg', '.png'))])
print(f"Loaded {len(non_agri_images)} non-agricultural images and {len(agri_images)} agricultural images.")
```

### Task 1: Determine the shape (dimensions) of a single image stored in the image_data variable (Question 1)


```python
image_data = plt.imread(os.path.join(dir_non_agri, non_agri_images[0]))
print("Image dimensions (shape):", image_data.shape)
```

### Task 2: Display the first four images in './images_dataSAT/class_0_non_agri/' directory (Question 2)


```python
n_images = 4
for image_path in non_agri_images[:n_images]:
    print(image_path)
    img = Image.open(os.path.join(dir_non_agri, image_path))
    plt.imshow(img)
    plt.show()
```

### Task 3: Create a list named agri_images_paths that contains the full file paths of all images located in the dir_agri directory. Sort the list before saving it.


```python
agri_images_paths = []
for image in os.listdir(dir_agri):
    agri_images_paths.append(os.path.join(dir_agri, image))

agri_images_paths.sort()
print(f"Total agricultural image paths saved: {len(agri_images_paths)}")
print("First path:", agri_images_paths[0])
```

### Task 4: Determine the number of images of agricultural land that exist in the './images_dataSAT/class_1_agri/' directory (Question 3)


```python
print(len(agri_images_paths))
```

### Task 5: Display the first four images of the agricultural land (Question 4)


```python
n_images = 4
for image_path in agri_images_paths[:n_images]:
    print(image_path)
    img = Image.open(image_path)
    plt.imshow(img)
    plt.show()
```
