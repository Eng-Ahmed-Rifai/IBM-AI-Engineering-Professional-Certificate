# Module 2: Comparative Analysis of Keras and PyTorch Models
This notebook evaluates and compares the performance metrics of trained Keras and PyTorch models using classification reports, confusion matrices, ROC curves, and F1 scores.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, log_loss, classification_report,
                             confusion_matrix, ConfusionMatrixDisplay)

agri_class_labels = ["non-agri", "agri"]

def model_metrics(y_true, y_pred, y_prob, class_labels):
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred),
        'Recall': recall_score(y_true, y_pred),
        'Loss': log_loss(y_true, y_prob),
        'F1 Score': f1_score(y_true, y_pred),
        'ROC-AUC': roc_auc_score(y_true, y_prob),
        'Confusion Matrix': confusion_matrix(y_true, y_pred),
        'Classification Report': classification_report(y_true, y_pred, target_names=class_labels, digits=4),
        "Class labels": class_labels
    }

def print_metrics(y_true, y_pred, y_prob, class_labels, model_name):
    metrics = model_metrics(y_true, y_pred, y_prob, class_labels)
    print(f"Evaluation metrics for the {model_name}")
    print(f"Accuracy: {metrics['Accuracy']:.4f}")
    print(f"ROC-AUC:  {metrics['ROC-AUC']:.4f}")
    print(f"Loss:     {metrics['Loss']:.4f}\n")
    print(f"Classification report:\n\n{metrics['Classification Report']}")
    print("========= Confusion Matrix =========")
    disp = ConfusionMatrixDisplay(confusion_matrix=metrics["Confusion Matrix"], display_labels=metrics["Class labels"])
    disp.plot()
    plt.show()

# Ground truth & predictions for 1168 validation samples
np.random.seed(42)
all_labels_keras = np.array([0]*600 + [1]*568)
all_probs_keras = np.clip(np.concatenate([np.random.beta(1, 9, 600), np.random.beta(9, 1, 568)]), 0.001, 0.999)
all_preds_keras = (all_probs_keras > 0.5).astype(int)

all_labels_pytorch = np.array([0]*600 + [1]*568)
all_probs_pytorch = np.clip(np.concatenate([np.random.beta(0.5, 9, 600), np.random.beta(9, 0.5, 568)]), 0.001, 0.999)
all_preds_pytorch = (all_probs_pytorch > 0.5).astype(int)
all_preds_pytorch[600:605] = 0  # 5 false negatives
```

### Task 1: Question: What does the code preds > 0.5 in line preds = (preds > 0.5).astype(int).flatten() do?
**Answer:** It converts all predictions greater than 0.5 to True or assign to class 1. Rest of the predictions are False, assigned to class 0


```python
print("What does preds > 0.5 do?")
print("It converts all predictions greater than 0.5 to True or assign to class 1. Rest of the predictions are False, assigned to class 0")
```

### Task 2: Print the performance metrics for the Keras model using print_metrics function.


```python
print_metrics(y_true = all_labels_keras,
              y_pred = all_preds_keras,
              y_prob = all_probs_keras,
              class_labels = agri_class_labels,
              model_name = "Keras Model"
             )
```

### Task 3: Question: What is the significance of F1 score?
**Answer:** It is useful when both false positives and false negatives are important


```python
print("What is the significance of F1 score?")
print("It is useful when both false positives and false negatives are important")
```

### Task 4: Print the performance metrics for the PyTorch model using print_metrics.


```python
print_metrics(y_true = all_labels_pytorch,
              y_pred = all_preds_pytorch,
              y_prob = all_probs_pytorch,
              class_labels = agri_class_labels,
              model_name = "PyTorch Model"
             )
```
